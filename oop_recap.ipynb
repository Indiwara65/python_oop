### Basics
* A class is like a blueprint to a data structure.
* Class is a blueprint and each instance of a class is called a instance of that class and will be according to that class/blueprint.

```python
class Employee():
    raise_amount = 1.04                    #class varaible constant across all instances
    num_of_emp = 0
    def __init__(self, firstname:str, lastname:str, pay:float):
        self.firstname = firstname         #instance variables
        self.lastname = lastname
        self.pay = pay
        self.email = f"{firstname.lower()}.{lastname.lower()}@topgear.original.com"

        Employee.num_of_emp += 1

    ##regular methods - takes a class instance as the first argument to the method (which is self in the below cases)
    def fullname(self):                   #class methods
        return f"{self.firstname} {self.lastname}"
    
    def raise_pay(self,amount:float|None=None):
        if amount is None:
            self.pay = self.pay * self.raise_amount    #when we use self.raise_amount then it looks for a instance varaiable then if not available to the class variable
            #if a instance varable names "raise_amount" is defined it will use that instance variable
            #if we want to use only the class variable then use Employee.raise_amount instead of self.raise_amount
        else:
            self.pay = self.pay + amount

    ##class methods - takes the class as the first argument to the method (which is cls in the below cases)
    @classmethod
    def set_raise_amt(cls, amount:float):
        cls.raise_amount = amount

    @classmethod
    def from_string(cls, emp_str:str):
        firstname, lastname, pay = emp_str.split("-")
        return cls(firstname, lastname, float(pay))
    
    ##static methods - it takes nither the class or a class instance as the first argument (similar to a normal function)
    @staticmethod
    def valid_pay(amount:float):
        return 20000<amount<15000
    
    ##special methods - __init__ is also a special method
    def __str__(self,):
        string = f"Employee Details\n-->First Name : {self.firstname}\n-->Last Name : {self.lastname}\n-->Email : {self.email}"
        return string

    def __repr__(self):
        string = f"{self.firstname}-{self.lastname}-{self.email}"
        return string
    
    def __add__(self, other):
        if isinstance(other, Employee):
            return self.pay + other.pay
        else:
            return NotImplementedError
            #return NotImplemented

    def __len__(self,):
        return len(self.fullname())
```

#### Inheritance           
* Create new functions with inherited methods and attributes from the parent class.          
* With class Inheritance we can create new classes with additional funtionality ither variables or methods

```python
class Developer(Employee):
    raise_amount = 1.07
    def __init__(self, firstname:str, lastname:str, pay:float, prog_lang:str):
        super().__init__(firstname, lastname, pay)   #parent init method
        self.prog_lang = prog_lang

class Manager(Employee):
    def __init__(self, firstname, lastname, pay, employees=None):          #never pass mutable data types as default arguments
        super().__init__(firstname, lastname, pay)
        self.employees = employees if employees is not None else []

    def add_emp(self, employee:Employee):
        if employee not in self.employees:
            self.employees.append(employee)

    def remove_emp(self, employee:Employee):
        if employee in self.employees:
            self.employees.remove(employee)

    def print_emp(self):
        print("------Employees------")
        for emp in self.employees:
            print(f"--->{emp.fullname()}")
```

```python
man_1 = Manager('Sigithi', 'Putha', 120000)
dev_1 = Developer('Cleetus','McFarland',100000,'python')
dev_1.email, dev_1.prog_lang
```

### Decoraters        
1. property decorator         
    * By adding the decorater @property over a method we can call that method similar to a variable without the parathesis (similar to calling a variable).
    * In the below case the *email* instance varaible is converted to a method.
2. setter
    * After the variable is converted into a method rather than a variable now the setter decorator can be used to define a function to set the variable.
    * In the below case by using the @email.setter decorater it is made possible to set the email as a varaible.   
3. deltter
    * Now a deleter can be set to the variable/method which can delete other attributes.
    * In the below case when the email is deleted the first name and last name will be deleted

```python
class Employee():
    def __init__(self, firstname:str, lastname:str, pay:float):
        self.firstname = firstname         
        self.lastname = lastname
        self.pay = pay

    def fullname(self):                  
        return f"{self.firstname} {self.lastname}"
    
    @property
    def email(self):
        return f"{self.firstname.lower()}.{self.lastname.lower()}@topgear.original.com"
    
    @email.setter
    def email(self, email:str):
        email_split = email.split('@')[0]
        firstname, lastname = email_split.split('.')
        self.firstname = f"{firstname[0].upper()}{firstname[1:]}"
        self.lastname = f"{lastname[0].upper()}{lastname[1:]}"

    @email.deleter
    def email(self,):
        self.firstname = None
        self.lastname = None
    
```

```python
emp = Employee('Jimmy','May',65000)
emp.email, emp.fullname()
```

```python
email = "james.may@topgear.original.com"
emp.email = email
emp.email, emp.fullname()
```

```python
del emp.email
emp.fullname()
```
